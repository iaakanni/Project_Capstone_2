```python
import pandas as pd
```

```python
file_path = r'C:\Users\isaac\OneDrive\Desktop\Project_Capstone 2\Project_Capstone_2\dataset\rawdata\Sales Dataset - Sales Dataset.csv'
```

```python
df = pd.read_csv(file_path)
```

```python
df.head()
```

```python
df.columns = df.columns.str.replace(" ", "_").str.strip()
df.columns = df.columns.str.replace("PaymentMode", "Payment_Mode").str.strip()
df.columns = df.columns.str.replace("CustomerName", "Customer_Name").str.strip()
df.columns = df.columns.str.replace("Year-Month", "Year_Month").str.strip()
df.info()
```

```python
df.head()
```

```python
# convert Order_Date to datetime\n"

df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Year-Month"] = df["Order_Date"].dt.to_period("M")
```

```python
df.head()
```

```python
#Reset to avoid not being continous
df = df.reset_index(drop=True)
df["Order_ID"] = "OID-" + (10001 + df.index).astype(str)
```

```python
df.head()
```

```python
# Create new CustomerID column
df["Customer_ID"] = range(1001, 1001 + len(df))
df["Product_ID"] = range(1001, 1001 + len(df))

# Preview
df.head()
```

```python
df = df.reset_index(drop=True)
df["Customer_ID"] = "CID-" + (10001 + df.index).astype(str)
df["Product_ID"] = "PID-" + (10001 + df.index).astype(str)
df["Location_ID"] = "LID-" + (10001 + df.index).astype(str)
df["Date_ID"] = "DID-" + (10001 + df.index).astype(str)
df.head()
```

```python
df.tail()
```

```python
# Derive year, month, quarter, date_id for dim_date
df["Year"]    = df["Order_Date"].dt.year
df["Month"]   = df["Order_Date"].dt.month
df["Quarter"] = df["Order_Date"].dt.quarter
```

```python
df.info()
```

```python
df.head()
```

```python
# Drop duplicates & nulls
df.drop_duplicates(inplace=True)
df.dropna(inplace=True)
```

```python
# ── 3. TRANSFORMATIONS ────────────────────────────────────────────────────────

# Monthly profit aggregation
monthly_profit = (
    df.groupby('Year_Month')['Profit']
    .sum()
    .reset_index()
    .rename(columns={'Profit': 'Total_Profit'})
    .sort_values('Year_Month')
)

# Top 10 customers by total amount
top_customers = (
    df.groupby('Customer_Name')['Amount']
    .sum()
    .reset_index()
    .rename(columns={'Amount': 'Total_Amount'})
    .sort_values('Total_Amount', ascending=False)
    .head(10)
)

# Quantity sold per category
qty_per_category = (
    df.groupby('Category')['Quantity']
    .sum()
    .reset_index()
    .rename(columns={'Quantity': 'Total_Quantity'})
    .sort_values('Total_Quantity', ascending=False)
)

# Cities with low profitability (below median profit)
city_profit = (
    df.groupby('City')['Profit']
    .sum()
    .reset_index()
    .rename(columns={'Profit': 'Total_Profit'})
)
median_profit = city_profit['Total_Profit'].median()
low_profit_cities = city_profit[city_profit['Total_Profit'] < median_profit].sort_values('Total_Profit')

# ── 4. WRITE TO CSV ───────────────────────────────────────────────────────────
df.to_csv('Cleaned_Sales.csv', index=False)
monthly_profit.to_csv('Monthly_Profit.csv', index=False)
top_customers.to_csv('Top_Customers.csv', index=False)
qty_per_category.to_csv('Qty_Per_Category.csv', index=False)
low_profit_cities.to_csv('Low_Profit_Cities.csv', index=False)

print("Done. Files written:")
print("  Cleaned_Sales.csv, Monthly_Profit.csv, Top_Customers.csv, Qty_Per_Category.csv, Low_Profit_Cities.csv")
```

```python
cleaned_sales    = pd.read_csv('cleaned_sales.csv')
monthly_profit   = pd.read_csv('monthly_profit.csv')
top_customers    = pd.read_csv('top_customers.csv')
qty_per_category = pd.read_csv('qty_per_category.csv')
low_profit_cities = pd.read_csv('low_profit_cities.csv')

print("=== Cleaned Sales ===")
print(cleaned_sales.head())

print("\n=== Monthly Profit ===")
print(monthly_profit)

print("\n=== Top Customers ===")
print(top_customers)

print("\n=== Quantity per Category ===")
print(qty_per_category)

print("\n=== Low Profit Cities ===")
print(low_profit_cities)
```
